```python
import pandas as pd

df = pd.read_csv(r'E:\FraudDetectionProject\data\raw\creditcard.csv\creditcard.csv')
df = df.sort_values('Time')

train_idx = int(len(df) * 0.70)
val_idx = int(len(df) * 0.85)

train = df.iloc[:train_idx]
val = df.iloc[train_idx:val_idx]
test = df.iloc[val_idx:]

train.to_csv(r'E:\FraudDetectionProject\data\processed\train.csv', index=False)
val.to_csv(r'E:\FraudDetectionProject\data\processed\val.csv', index=False)
test.to_csv(r'E:\FraudDetectionProject\data\processed\test.csv', index=False)
```
